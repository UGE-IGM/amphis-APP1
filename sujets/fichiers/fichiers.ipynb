# Lecture de fichiers



Il est très facile de consulter le contenu d'un fichier texte en Python, et de le modifier ([doc officielle Python](https://docs.python.org/fr/3/tutorial/inputoutput.html#reading-and-writing-files))

Ouverture d'un fichier : `f = open(chemin, mode)`

- Paramètre `chemin` (chaîne de caractère) :  
  nom (absolu ou relatif) d'un fichier
- Paramètre `mode` (chaîne de caractère) :
  - lecture seule : `'r'` (par défaut)
  - écriture (crée / écrase le fichier) : `'w'`
  - ajout (écrit à la fin) : `'a'`
  - *autres modes possibles pour fichiers binaires*

Fermeture : `f.close()`

Lecture (mode `'r'`) :

-   Tout lire d'un coup (une seule chaîne) :  
    `contenu = f.read()`  
    **Attention :** charge tout le fichier en mémoire !  
    **Attention :** inclut les retours à la ligne

```python
f = open('test.txt', 'r')
f.read()
```

-   Liste des lignes (liste de chaînes) :  
    `lst_lignes = f.readlines()`  
    **Attention :** charge tout le fichier en mémoire !  
    **Attention :** inclut les retours à la ligne

```python
f = open('test.txt', 'r')
f.readlines()
```

    
-   Prochaine ligne : `ligne = f.readline()`  
    **Attention :** renvoie une chaîne vide à la fin
    **Attention :** inclut les retours à la ligne

```python
f = open('test.txt', 'r')
f.readline()
```

```python
f.readline()
```

```python
f.readline()
```

```python
f.readline()
```

-   Parcours ligne par ligne : `for ligne in f: ...`  
    **Attention :** inclut les retours à la ligne

```python
f = open('test.txt', 'r')
for ligne in f:
    print(repr(ligne))
f.close()
```

## Fichiers binaires

On peut spécifier que le fichier que l'on ouvre doit être interprété comme un fichier *binaire* (c'est à dire octet par octet) en ajoutant le caractère 'b' au mode d'ouverture. Par exemple :

```python
f = open('guybrush3.p6.ppm', mode='rb')
```

On peut lire le contenu du fichier comme d'habitude :

```python
format, width, height, depth, pixmap = f.read().split()
```

... mais on obtient des chaînes binaires, et non des chaînes "normales", c'est à dire que le contenu est interprété octet par octet :

```python
format, type(format)
```

```python
width, height, depth
```

```python
pixmap, len(pixmap) % 3
```

Contrairement à une chaîne classique, l'accès à un élément d'une chaîne binaire renvoie une valeur de type `int` : c'est la valeur de l'entier codé par l'octet correspondant.

```python
format[0], type(format[0]), chr(format[0])  # c'est aussi un 'P' !
```

```python
pixmap[0], type(pixmap[0])
```

Si l'on accède à plusieurs valeurs d'un coup avec une tranche, on obtient à nouveau une chaîne binaire. Il faut donc convertir chaque élément séparément.

```python
tuple(map(int, pixmap[0:3]))
```

```python
image = []
ligne = []
width = int(width)
for i in range(0, len(pixmap), 3):
    pixel = tuple(map(int, pixmap[i:i+3]))
    ligne.append(pixel)
    if len(ligne) == width:
        image.append(ligne)
        ligne = []
image
```

```python
len(image) == int(height)
```

```python
len(pixmap) == width * int(height) * 3
```
