# Exercice sur les listes : ensembles finis



## Type de données abstrait



On souhaite créer des structures de données en Python et les fonctions associées pour manipuler des **ensembles** de valeurs (supposés de même type, et *comparables* deux à deux). On veut pouvoir :

- créer un ensemble vide : `ensemble_vide()`
- ajouter un élément à un ensemble : `ajoute(ensemble, elem)`
- déterminer si un ensemble contient un élément donné : `contient(ensemble, elem)`
- déterminer le cardinal (nombre d'éléments) d'un ensemble : `cardinal(ensemble)`

De plus, on voudrait pouvoir comparer et combiner des ensembles entre eux :

- déterminer si un ensemble est inclus dans un autre : `inclus_dans(ens1, ens2)`
- calculer la réunion de deux ensembles : `union(ens1, ens2)`
- calculer l'intersection de deux ensembles : `intersection(ens1, ens2)`
- calculer la différence de deux ensembles : `difference(ens1, ens2)`

On appelle ce genre de spécification un **type abstrait de données** (abstrait parce qu'on ne précise pas comment on l'implémente !)



## Idées d'implémentation



Quelles implémentations possibles ?

- Avec une liste (ordinaire) sans doublon
- Avec une liste **triée** sans doublon

**En Python :** type `set`, *tables de hachage* (avec plein d'optimisations)

### Listes non triées

```python
def ensemble_vide():
    return []
```

```python
def ajoute(ensemble, elem):
    if elem not in ensemble:
        ensemble.append(elem)
```

```python
def contient(ensemble, elem):
    return elem in ensemble
```

```python
def cardinal(ensemble):
    return len(ensemble)
```

```python
ens = ensemble_vide()
ajoute(ens, 1)
ajoute(ens, 1)
ajoute(ens, 1)
ajoute(ens, 1)
print(contient(ens, 1), contient(ens, 2), cardinal(ens))
```

```python
def inclus_dans(un, deux):
    # renvoie True si l'ensemble un est inclus dans deux, False sinon
    for elem in un:
        if elem not in deux:  # coûteux !
            return False
    return True
```

```python
inclus_dans([4, 2, 7], [9, 1, 2, 3, 4, 5, 7, 6])
```

```python
inclus_dans([4, 2, 11], [9, 1, 2, 3, 4, 5, 7, 6])
```

```python
def union(un, deux):
    res = un.copy()  # coûteux
    for elem in deux:
        ajoute(res, elem)  # coûteux !!
    return res
```

```python
union([4, 2, 11], [9, 1, 2, 3, 4, 5, 7, 6])
```

```python
def intersection(un, deux):
    res = ensemble_vide()
    for elem in un:
        if elem in deux:  # coûteux
            res.append(elem)  # coûteux
    return res
```

```python
intersection([4, 2, 11], [9, 1, 2, 3, 4, 5, 7, 6])
```

```python
def difference(un, deux):
    res = ensemble_vide()
    for elem in un:
        if elem not in deux:  # coûteux
            res.append(elem)
    return res
```

```python
difference([4, 2, 11], [9, 1, 2, 3, 4, 5, 7, 6])
```
