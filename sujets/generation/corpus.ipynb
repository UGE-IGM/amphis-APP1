# Représentation du corpus



```python
exemple = {
    "le": {
        "petit": {
            "chat": {"blanc": 2, "noir": 1, "mange": 1},
            "chien": {"blanc": 1, "aboie":2},
            "garçon": {"joue": 1, "mange": 1, "aboie": 1}
        },
        "grand": {
            "chien": {"joue": 1, "aboie":1}
        }
    }
}
```

```python
def inserer(corpus, n_gramme, mot_suivant):
    # je localise mon dictionnaire d'occurrences
    for mot in n_gramme:
        if mot not in corpus:
            corpus[mot] = {}
        corpus = corpus[mot]
    # j'inscris une occurrence de mot_suivant
    if mot_suivant in corpus:
        corpus[mot_suivant] += 1
    else:
        corpus[mot_suivant] = 1
```

```python
inserer(exemple, ["le", "petit", "chat"], "blanc")
inserer(exemple, ["le", "petit", "chat"], "perdu")
exemple
```

```python
phrases = [
    "A 01h10, les combattants de la Résistance islamique ont frappé pour la deuxième fois un rassemblement de soldats ennemis israéliens à proximité de la localité d’Aita al-Shaab alors qu’ils tentaient d’évacuer les soldats tués et blessés.",
    "A 11 ans, il intègre le conservatoire de Bordeaux et commence à suivre le collège à domicile, avant d’entrer au lycée Camille-Jullian de Bordeaux, en classe musicale.",
    "À 11 heures, il lui envoyait encore des photos avec son smartphone.",
    "À 16 ans, elle a dû arrêter sa carrière à cause d'une grave déchirure du ligament croisé antérieur.",
    "A 17h50, ce sont les quatre marchés de Noël de la capitale (Lëtzebuerger Chrëschtmaart, Wanterpark, Niklosmaart, Wantermaart) qui ont été fermés.",
    "À 19h30, Hydro-Québec signalait qu’un peu plus de 100 000 clients étaient encore privés d’électricité, dont 34 095 dans les Laurentides et près de 5000 à Montréal.",
    "À 1 minute 54 secondes, Alexis Lemire repère Evan Courtois dans l’enclave qui réussit à marquer un premier but pour la meute avec son neuvième de la saison.",
    "À 24 ans, soit presque dix ans après la mort de sa mère des suites d’un cancer, Paul McCartney a décidé, inconsciemment, de faire de ce regret un morceau.",
    "À 52 ans, après avoir quitté la Patinoire royale, j'ai eu envie que les prochaines années puissent donner libre cours à ma passion, mes convictions et mes engagements culturels et artistiques, en toute indépendance.",
    "À 80 $ la tonne, le prix du carbone ajoutera 15,3 cents au mètre cube de gaz naturel, comparativement à 12,4 cents auparavant.",
    "À 83 ans, Michel est particulièrement en forme : tel un bon vin, il ne vieillit pas., nous lance-t-il, sourire en coin, depuis son fauteuil dans son appartement parisien près de Saint-Germain-des-Prés.",
    "À allure modérée, nous quittons Hal en remontant le Ring, direction le centre de la capitale.",
    "À aucun moment dans cet article, on lit la vraie raison de cette bataille."
]
```

```python
def construire(phrases, n):
    phrases = nettoyer(phrases)
    corpus = {}
    for phrase in phrases:
        for i in range(len(phrase) - n):
            inserer(corpus, phrase[i:n+i], phrase[n+i])
    return corpus

corpus = construire(phrases, 1)
```

```python
import string

def nettoyer(phrases):
    trans = str.maketrans({punct: " " for punct in string.punctuation})
    return [phrase.translate(trans).lower().strip().split() for phrase in phrases]
```

```python
def completer(corpus, n_gramme):
    courant = corpus
    for mot in n_gramme:
        if mot not in courant:
            return {}  # à raffiner !?
        courant = courant[mot]
    return courant
```

```python
completer(corpus, ["le"])
```

```python
completer(corpus, ["centre"])
```

```python
corpus = construire(phrases, 2)
completer(corpus, ["le", "centre"])
```
